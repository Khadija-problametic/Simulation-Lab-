A customer service center studies how long customers wait before being served.
They suspect:
On regular days, customers arrive randomly and wait times are uniformly distributed
between 2 and 12 minutes.
On busy days, wait times follow a normal distribution with an average (mean) of
8 minutes and a standard deviation of 2 minutes.

```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

Part 1 — Data Generation
1. Generate 3,000 wait times using a uniform distribution between 2 and 12 minutes.
2. Generate 3,000 wait times using a normal distribution with:
o Mean = 8
o Std Dev = 2
3. Ensure no negative wait times in the normal distribution (replace negatives with 0).

```python
random_numbers=np.random.rand(30000)
#unifrom nubers
regular_days = np.random.uniform(2, 12, 3000)
#noirmal numbers

busy_wait = np.random.normal(8, 2, 3000)


busy_wait = np.clip(busy_wait, 0, None)

print("Regualr days : ",regular_days)
print("Busy day : ",busy_wait)
```

🔹 Part 2 — Visualization
1. Plot histograms for both distributions.
2. Show them:
o In the same figure (two subplots), OR
o Overlaid with transparency.

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

axes[0].hist(regular_days, bins=40, alpha=0.4, color='blue', edgecolor='black')
axes[0].set_title('Regular Day Wait Times Uniform')
axes[0].set_xlabel('Wait Time minutes')
axes[0].set_ylabel('Frequency')
axes[0].grid(True)

axes[1].hist(busy_wait, bins=50, alpha=0.5, color='red', edgecolor='black')
axes[1].set_title('Busy Day Wait Times Normal')
axes[1].set_xlabel('Wait Time minutes')
axes[1].set_ylabel('Frequency')
axes[1].grid(True)


plt.tight_layout()
plt.show()
```

Part 3 — Statistical Analysis
For both datasets calculate:
1. Mean
2. Median
3. Standard deviation
4. Maximum wait time

```python
 #Regular day
regular_mean = np.mean(regular_days)
regular_median = np.median(regular_days)
regular_std = np.std(regular_days)
regular_max = np.max(regular_days)

# Busy day
busy_mean = np.mean(busy_wait)
busy_median = np.median(busy_wait)
busy_std = np.std(busy_wait)
busy_max = np.max(busy_wait)

print("\nRegular Day :")
print("Mean:", regular_mean)
print("Median:", regular_median)
print("Standard Deviation:", regular_std)
print("Maximum Wait Time:", regular_max)

print("\nBusy Day :")
print("Mean:", busy_mean)
print("Median:", busy_median)
print("Standard Deviation:", busy_std)
print("Maximum Wait Time:", busy_max)
```
