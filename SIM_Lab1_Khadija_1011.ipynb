TASK-1: Scalar and *Array* Operations

```python
# Task 1: Scalar & Array Operations
scalar = 3.7
arr = np.array([4.2, 5.5, 6.8, 7.4, 8.1, 9.3])

# Trigonometric functions
sin_values = np.sin(arr)
cos_values = np.cos(arr)
tan_values = np.tan(arr)

# Inverse trig functions (clip array values into range [-1, 1] to prevent NaN/errors)
asin_values = np.arcsin(np.clip(arr / np.max(arr), -1, 1))
acos_values = np.arccos(np.clip(arr / np.max(arr), -1, 1))
atan_values = np.arctan(arr)

# Exponential and Natural Logarithm
exp_values = np.exp(arr)
log_values = np.log(arr)

# Absolute value and Square root
abs_values = np.abs(arr)
sqrt_values = np.sqrt(arr)

# Remainder when divided by scalar
rem_values = np.remainder(arr, scalar)

# Rounding functions
round_values = np.round(arr)
floor_values = np.floor(arr)
ceil_values = np.ceil(arr)

# Print results
print("###### Task 1 Outputs ######")
print("Scalar:", scalar)
print("Array:", arr)
print("sin:", sin_values)
print("cos:", cos_values)
print("tan:", tan_values)
print("asin:", asin_values)
print("acos:", acos_values)
print("atan:", atan_values)
print("exp:", exp_values)
print("log:", log_values)
print("abs:", abs_values)
print("sqrt:", sqrt_values)
print("rem (divided by scalar):", rem_values)
print("round:", round_values)
print("floor:", floor_values)
print("ceil:", ceil_values)
```

Task-2:Operations on a 3×2 Matrix


```python
# Task 2: 3x2 Matrix Operations
matrix_3x2 = np.array([
    [1.5, 2.8],
    [3.2, 4.9],
    [5.1, 6.0]
])

# Trigonometric functions
sin_m = np.sin(matrix_3x2)
cos_m = np.cos(matrix_3x2)
tan_m = np.tan(matrix_3x2)

# Inverse trig functions with domain clipping
asin_m = np.arcsin(np.clip(matrix_3x2 / np.max(matrix_3x2), -1, 1))
acos_m = np.arccos(np.clip(matrix_3x2 / np.max(matrix_3x2), -1, 1))
atan_m = np.arctan(matrix_3x2)

# Exponential and Logarithm
exp_m = np.exp(matrix_3x2)
log_m = np.log(matrix_3x2)

# Absolute value and Square root
abs_m = np.abs(matrix_3x2)
sqrt_m = np.sqrt(matrix_3x2)

# Remainder when divided by 2
rem_m = np.remainder(matrix_3x2, 2)

# Rounding
round_m = np.round(matrix_3x2)
floor_m = np.floor(matrix_3x2)
ceil_m = np.ceil(matrix_3x2)

print("###### Task 2 Outputs ######")
print("Original 3x2 Matrix:\n", matrix_3x2)
print("\nsin:\n", sin_m)
print("\ncos:\n", cos_m)
print("\ntan:\n", tan_m)
print("\nasin:\n", asin_m)
print("\nacos:\n", acos_m)
print("\natan:\n", atan_m)
print("\nexp:\n", exp_m)
print("\nlog:\n", log_m)
print("\nabs:\n", abs_m)
print("\nsqrt:\n", sqrt_m)
print("\nrem (divided by 2):\n", rem_m)
print("\nround:\n", round_m)
print("\nfloor:\n", floor_m)
print("\nceil:\n", ceil_m)
```

Task-3:Vector Functions on 2x3 Matrix

```python
import numpy as np

# Sample 2x3 matrix with floating-point numbers
matrix_2x3 = np.array([
    [10.2, 13.2, 12.4],
    [21.7, 16.1, 11.9]
])

# 1. Maximum value and index of max element
max_value = np.max(matrix_2x3)
max_index = np.argmax(matrix_2x3)  # Flat index of max element

# 2. Minimum value and index of min element
min_value = np.min(matrix_2x3)
min_index = np.argmin(matrix_2x3)  # Flat index of min element

# 3. Length of the matrix (total number of elements)
matrix_length = len(matrix_2x3.flatten())  # Equivalent to total length

# 4. Sorting in descending order
# Row-wise descending sort
row_sorted_desc = -np.sort(-matrix_2x3, axis=1)

# Entire matrix descending sort
entire_sorted_desc = -np.sort(-matrix_2x3.flatten())

# 5. Sum of elements
sum_values = np.sum(matrix_2x3)

# 6. Product of elements
prod_values = np.prod(matrix_2x3)

# 7. Median value
median_value = np.median(matrix_2x3)

# 8. Mean value
mean_value = np.mean(matrix_2x3)

# 9. Standard deviation
std_dev = np.std(matrix_2x3)

# Display results
print("###### Task 3 Outputs ######")
print("2x3 Matrix:\n", matrix_2x3)
print("\nMax value:", max_value, "at index", max_index)
print("Min value:", min_value, "at index", min_index)
print("Length of matrix (total elements):", matrix_length)
print("Row-wise sorted (descending):\n", row_sorted_desc)
print("Entire matrix sorted (descending):", entire_sorted_desc)
print("Sum of elements:", sum_values)
print("Product of elements:", prod_values)
print("Median value:", median_value)
print("Mean value:", mean_value)
print("Standard deviation:", std_dev)
```
