```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

TASK-1. Create Your Personal Matrix


```python
d1=1
d2=1
A = np.array([[d1+2 , d2+1 ], [2*d1, d2+2]])#createing matrix
print("my matrix:\n",A)
```

TASK-2. Analyze Matrix A

```python
# 1. Size of a matrix
size_A = A.shape

# 2. Determinant of a matrix
det_A = np.linalg.det(A)

# 3. Inverse of a matrix
if np.linalg.det(A) != 0:  # Check if matrix is invertible
    inv_A = np.linalg.inv(A)  # Compute the inverse of A
else:
    inv_A = "Matrix is singular, cannot compute inverse"  # No inverse if determinant is 0

# 4. Rank of a matrix
rank_A = np.linalg.matrix_rank(A)  # Returns the rank of the matrix

# 5. Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)  # Compute eigenvalues and eigenvectors




# Display results
print("Size of A:", size_A)

print("Determinant of A:", det_A)

print("Rank of A:", rank_A)

print("Eigenvalues of A:", eigenvalues)

print("Inverse of A:\n", inv_A)
```

TASK3. Value Change Experiment

```python
B = A.copy()
B[1,0] += 1
# 1. Size of a matrix
size_B = B.shape

# 2. Determinant of a matrix
det_B = np.linalg.det(B)

# 3. Inverse of a matrix
if np.linalg.det(B) != 0:  # Check if matrix is invertible
    inv_B = np.linalg.inv(B)  # Compute the inverse of A
else:
    inv_B= "Matrix is singular, cannot compute inverse"  # No inverse if determinant is 0

# 4. Rank of a matrix
rank_B = np.linalg.matrix_rank(B)  # Returns the rank of the matrix

# 5. Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(B)  # Compute eigenvalues and eigenvectors




# Display results
print("Size of B:", size_B)

print("Determinant of B:", det_B)

print("Rank of B:", rank_B)

print("Eigenvalues of B:", eigenvalues)

print("Inverse of B:\n", inv_B)
```

1. How did the determinant change and why?

ANSWER: The determinant decreased by 2, dropping from 5 in Matrix $A$ to 3 in Matrix $B$.This happens because of the $2 \times 2$ determinant formula $\det = (a \cdot d) - (b \cdot c)$. I added $+1$ to the bottom-left element ($c$, position [1,0]). In the formula, $c$ is subtracted after being multiplied by the top-right element ($b = 2$). Adding $+1$ to $c$ subtracts an extra $1 \times b = 2$ from the total, bringing the determinant down from 5 to 3 ($5 - 2 = 3$).

2. Did the rank change? Explain.

ANSWER:No, the rank stayed at 2 for both matrices.A $2 \times 2$ matrix has a full rank of 2 as long as its determinant is non-zero (meaning its rows are linearly independent). Matrix $A$ has a determinant of 5 and Matrix $B$ has a determinant of 3. Since neither determinant is zero, neither matrix is singular, and both retain a full rank of 2.

3. How did the eigenvalues respond to the value change?

ANSWER:The eigenvalues spread further apart, moving from $[5, 1]$ in $A$ to approximately $[5.30, 0.70]$ in $B$.
Sum equals Trace: Since no main diagonal elements were modified, the trace (diagonal sum) remained $3 + 3 = 6$. Notice that $5.3028 + 0.6972 = 6$ . The product of the new eigenvalues equals the new determinant ($5.3028 \times 0.6972 \approx 3$).To keep the diagonal sum at 6 while lowering the determinant to 3, one eigenvalue had to increase while the other decreased.

4. Is B easier or harder to invert than A?

ANSWER:Matrix $B$ is hard to invert than Matrix $A$.Computing a matrix inverse involves dividing by the determinant. The closer a determinant gets to zero, the more sensitive the calculation becomes to numerical rounding errors. Because Matrix $B$ has a smaller determinant magnitude ($3$) than Matrix $A$ ($5$), it is closer to being singular ($\det = 0$), making it slightly more ill-conditioned and harder to invert accurately.
